# AzuDL GUI · Drive edition

Run both cells, then authorize Google Drive when prompted. Direct files, GitHub assets, and media intermediates are stored on Drive.

```python
from pathlib import Path
import os
import shutil
import subprocess
import sys

project = Path("/content/azudl-gc2gd-drive")
if not project.exists():
    subprocess.run([
        "git", "clone", "--depth", "1",
        "https://github.com/asmitsharma-alt/AzuDL-GC2GD.git", str(project)
    ], check=True)
subprocess.run([
    sys.executable, "-m", "pip", "install", "-q", "yt-dlp[default]==2026.11.23"
], check=True)

if not shutil.which("deno"):
    tools = Path("/content/azudl-drive-tools")
    deno_bin = tools / "node_modules" / ".bin"
    if not (deno_bin / "deno").exists():
        subprocess.run(["npm", "install", "--prefix", str(tools), "--no-audit", "deno@2.9.6"], check=True)
    os.environ["PATH"] = str(deno_bin) + os.pathsep + os.environ["PATH"]
```
